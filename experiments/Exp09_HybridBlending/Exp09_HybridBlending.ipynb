# Exp09: Hybrid Blending (Exp07 + Exp08)
## Deskripsi
Eksperimen ini menggabungkan model Tree-Based Ratio Targeting (Exp07) dan Deep Learning Entity Embeddings (Exp08).
Formula Blending:
`submission['total_ticket'] = 0.50 * sub_exp07['total_ticket'] + 0.50 * sub_exp08['total_ticket']`
## Hasil
Public LB Score: 0.7133. Mengonfirmasi bahwa batas irreducible error pohon keputusan mentah adalah 0.6252 (Exp07).


```python
import pandas as pd
import numpy as np

sub07 = pd.read_csv('experiments/Exp07_RatioTargeting_FeaturePruning/submission.csv')
sub08 = pd.read_csv('experiments/Exp08_DeepLearningEmbeddings/submission.csv')

sub09 = sub07.copy()
sub09['total_ticket'] = np.round(0.50 * sub07['total_ticket'] + 0.50 * sub08['total_ticket'])
sub09.to_csv('experiments/Exp09_HybridBlending/submission.csv', index=False)
print("Exp09 Hybrid Blending generated successfully.")
```
